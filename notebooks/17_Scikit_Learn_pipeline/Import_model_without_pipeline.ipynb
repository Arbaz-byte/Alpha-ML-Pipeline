```python
import pickle
import numpy as np
import pandas as pd
```

```python
ohe_sex = pickle.load(open('../Utilities/Model/ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open('../Utilities/Model/ohe_embarked.pkl', 'rb'))
clf = pickle.load(open('../Utilities/Model/clf.pkl', 'rb'))
```

```python
# User Input : Pclass/gender/age/SibSp/Parch/Fare/Embarked
test_input =np.array([2, 'male', 31.0, 0, 0, 11, 'S'], dtype=object).reshape(1,7)
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
test_input_age = test_input[:,2].reshape(1,1)


sex_cols = ohe_sex.get_feature_names_out(['Sex']).tolist()
embarked_cols = ohe_embarked.get_feature_names_out(['Embarked']).tolist()

cols = ['Pclass', 'SibSp', 'Parch', 'Fare', 'Age'] + sex_cols + embarked_cols
test_input_transformed = pd.DataFrame(
    np.concatenate((test_input[:,[0,3,4,5]], test_input_age, test_input_sex, test_input_embarked),axis=1),
    columns=cols

)
```

```python
clf.predict(test_input_transformed)
```
