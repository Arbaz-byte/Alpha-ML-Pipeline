# Principal Component Analysis (PCA) 

Principal Component Analysis (PCA) is an unsupervised linear transformation technique used for dimensionality reduction. Its goal is to find a new set of orthogonal axes—called **Principal Components (PCs)**—that maximize the variance of the projected data while minimizing information loss.

---

## 1. Complete Mathematical Derivation

Let our raw dataset $\mathbf{X}_{\text{raw}}$ be an $n \times d$ matrix, where $n$ is the number of samples and $d$ is the number of features.

### Step 1: Mean Centering and Standardization
To ensure that PCA aligns with the internal variance of the data cloud rather than its absolute position relative to the origin, we compute the empirical row mean vector $\boldsymbol{\mu} = [\mu_1, \mu_2, \dots, \mu_d]$ where:
$$\mu_j = \frac{1}{n} \sum_{i=1}^{n} x_{i,j}$$

The mean-centered data matrix $\mathbf{X}$ is obtained by subtracting the mean from each row:
$$\mathbf{X} = \mathbf{X}_{\text{raw}} - \mathbf{1}_n \boldsymbol{\mu}$$
*(Where $\mathbf{1}_n$ is an $n \times 1$ column vector of ones).*

To eliminate scale dependencies across mismatched units, we optionally scale each column by its standard deviation $\sigma_j$:
$$x_{i,j(\text{standardized})} = \frac{x_{i,j} - \mu_j}{\sigma_j}$$

---

### Step 2: Defining the Projection Framework
We want to project our data matrix $\mathbf{X}$ onto a new directional axis defined by a column vector $\mathbf{w}$ of size $d \times 1$. To prevent arbitrary scaling of the projections, we enforce that $\mathbf{w}$ must be a unit vector:
$$\|\mathbf{w}\|_2 = 1 \implies \mathbf{w}^T\mathbf{w} = 1$$

The scalar projection of all data points onto this axis creates an $n \times 1$ coordinate vector $\mathbf{y}$:
$$\mathbf{y} = \mathbf{X}\mathbf{w}$$

---

### Step 3: Formulating the Variance Objective Function
Because $\mathbf{X}$ is mean-centered, the mean of the projected coordinates $\mathbf{y}$ is also zero ($\mathbb{E}[\mathbf{y}] = 0$). The empirical variance of our projections along $\mathbf{w}$ is defined as:
$$\text{Variance}(\mathbf{y}) = \frac{1}{n} \sum_{i=1}^n y_i^2 = \frac{1}{n} \mathbf{y}^T \mathbf{y}$$

Substituting $\mathbf{y} = \mathbf{X}\mathbf{w}$ into the equation:
$$\text{Variance}(\mathbf{y}) = \frac{1}{n} (\mathbf{X}\mathbf{w})^T (\mathbf{X}\mathbf{w})$$
Using the transposition property $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:
$$\text{Variance}(\mathbf{y}) = \frac{1}{n} \mathbf{w}^T \mathbf{X}^T \mathbf{X} \mathbf{w} = \mathbf{w}^T \left( \frac{1}{n} \mathbf{X}^T \mathbf{X} \right) \mathbf{w}$$

The term inside the parentheses is exactly the $d \times d$ **Sample Covariance Matrix** $\mathbf{\Sigma}$:
$$\mathbf{\Sigma} = \frac{1}{n} \mathbf{X}^T \mathbf{X}$$

Thus, our constrained optimization problem becomes:
$$\max_{\mathbf{w}} \quad \mathbf{w}^T \mathbf{\Sigma} \mathbf{w} \quad \text{subject to} \quad \mathbf{w}^T\mathbf{w} = 1$$

---

### Step 4: Solving via Lagrange Multipliers
To solve this constrained optimization, we introduce the Lagrange multiplier $\lambda$ and establish the scalar Lagrangian function $L(\mathbf{w}, \lambda)$:
$$L(\mathbf{w}, \lambda) = \mathbf{w}^T \mathbf{\Sigma} \mathbf{w} - \lambda (\mathbf{w}^T\mathbf{w} - 1)$$

We find the stationary points by calculating the gradient with respect to vector $\mathbf{w}$ and setting it to the zero vector $\mathbf{0}$:
$$\nabla_{\mathbf{w}} L(\mathbf{w}, \lambda) = 2\mathbf{\Sigma}\mathbf{w} - 2\lambda\mathbf{w} = \mathbf{0}$$
$$\mathbf{\Sigma}\mathbf{w} = \lambda\mathbf{w}$$

This is the classic **Eigenvalue Problem**. It explicitly proves that the optimal direction vectors $\mathbf{w}$ are the **eigenvectors** of the covariance matrix $\mathbf{\Sigma}$, and their corresponding **eigenvalues** $\lambda$ represent the exact variance captured along those axes.

---

### Step 5: Finding Subsequent Orthogonal Components
Since $\mathbf{\Sigma}$ is a symmetric real matrix, the Spectral Theorem guarantees it yields $d$ real eigenvalues sorted in descending order:
$$\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_d$$

To find the second principal component $\mathbf{w}_2$, we maximize $\mathbf{w}_2^T \mathbf{\Sigma} \mathbf{w}_2$ subject to $\mathbf{w}_2^T\mathbf{w}_2 = 1$ and the orthogonality constraint $\mathbf{w}_2^T\mathbf{w}_1 = 0$. The resulting Lagrangian proves that $\mathbf{w}_2$ is simply the eigenvector associated with the second largest eigenvalue $\lambda_2$.

---

### Step 6: Dimensionality Reduction Projection
To reduce our space from $d$ dimensions down to a target size $k$ ($k < d$), we build a $d \times k$ **Projection Matrix** $\mathbf{W}_k$ using the top $k$ eigenvectors:
$$\mathbf{W}_k = \begin{bmatrix} \vert & \vert & & \vert \\ \mathbf{w}_1 & \mathbf{w}_2 & \dots & \mathbf{w}_k \\ \vert & \vert & & \vert \end{bmatrix}$$

We multiply the original centered data matrix by $\mathbf{W}_k$ to yield the final compressed dataset:
$$\mathbf{X}_{\text{reduced}} = \mathbf{X}\mathbf{W}_k \quad \longrightarrow \quad [n \times d] \times [d \times k] = [n \times k]$$






## 2. Practical Applications & Usecases

* **Data Visualization:** Compresses complex datasets with hundreds of features down to a 2D or 3D scatter plot to identify structural clusters, categories, or anomalies visually.

* **Accelerating Machine Learning:** Curbs the *Curse of Dimensionality* by stripping highly correlated or redundant variables. This minimizes model training times and cuts computational memory usage.

* **Noise Filtering:** By discarding components tied to small eigenvalues, PCA acts as a low-pass filter, retaining signal patterns while purging random sensor or environmental noise.

* **Feature Preprocessing:** Used heavily in genomics (microarray tracking), image compression (Eigenfaces), and financial quantitative analysis to extract uncorrelated structural indicators.





## 3. Core Limitations

* **Linear Assumptions Only:** PCA assumes data clusters change linearly. If your underlying data structure is highly non-linear (e.g., Swiss roll shapes or concentric rings), standard PCA will flatten out critical structural information. Non-linear alternatives like *Kernel PCA* or *t-SNE* are required here.

* **Scale Sensitivity:** PCA is deeply sensitive to feature scaling. If you fail to apply Z-score standardization before processing, features with massive raw units will artificially dictate the principal components.

* **Loss of Interpretability:** Transformed features ($PC_1, PC_2$) are linear combinations of *all* original features. While you retain statistical variance, you lose the direct physical meaning of your initial features (e.g., you can no longer easily say exactly how much "Age" or "Income" alone impacted a specific model decision).

* **Outlier Vulnerability:** Because the variance calculation relies heavily on squared distances ($\frac{1}{n}\mathbf{X}^T\mathbf{X}$), extreme anomalies or outliers can heavily skew the direction of the principal axes away from the true majority data spread.
