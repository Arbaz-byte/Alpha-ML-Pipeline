```python
import pandas as pd
import numpy as np
```

```python
# Import the data from a CSV file
date = pd.read_csv('../data/orders.csv')
time = pd.read_csv('../data/messages.csv')
```

## Working On data

```python
date.head()
```

```python
date.drop(columns=['product_id', 'city_id', 'orders'], inplace=True)
date.head()
```

```python
date.info() # date is pandas object , we need to convert it into datetime format
```

```python
# Convert to datetime datatype
date['date'] = pd.to_datetime(date['date'])
date.info()
```

### 1. Extract Year

```python
date['date_year'] = date['date'].dt.year
date.head()
```

### 2. Extract Month

```python
date['Month_num'] = date['date'].dt.month
date.head()
```

```python
date['Month_name'] = date['date'].dt.month_name()
date.head()
```

### 3. Extract Day

```python
date['Day'] = date['date'].dt.day
date.head()
```

```python
# Day of week
date['Day_of_week'] = date['date'].dt.dayofweek
date.head()
```

```python
# Day of week (name)
date['Day_of_week_name'] = date['date'].dt.day_name()
date.head()
```

```python
# Is weekend ?

date['Isweekned'] = np.where(date['Day_of_week_name'].isin(['Sunday', 'Saturday']), 1, 0)
date.head()
```

### 4. Extract Week Of Year

```python
date['Week_of_year'] = date['date'].dt.isocalendar().week
date.head()
```

### 5. Extract Quarter

```python
date['Quarter'] = date['date'].dt.quarter
date.head()
```

### 6. Extract Semester


```python

date['Semester'] = np.where(date['Quarter'].isin([1,2]), 'First', 'Second')
date.head()
```

### 7. Extract Time elapsed b/w dates


```python
import datetime
today = datetime.datetime.today()
today
```

```python
diff_date = (today - date['date'])
diff_date
```

```python
# Month Passed
np.round((today - date['date']).dt.days / 30.44, 0)
```

## Working On Time 

```python
time.drop(columns=['msg'], inplace=True)
time.head()
```

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
time.info()
```

### 1. Extract hour, min and sec

```python
time['time'] = time['date'].dt.time
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second
time['Date'] = time['date'].dt.date
time.head()
```

### 2. Time Difference


```python
# In Seconds
(today - time['date'])/ np.timedelta64(1, 's')
```

```python
# In Minutes
(today - time['date']) / np.timedelta64(1, 'm')
```

```python
# In Hours 
(today - time['date']) / np.timedelta64(1, 'h')
```
